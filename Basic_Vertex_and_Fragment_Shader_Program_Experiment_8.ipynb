```python
# Experiment 8: Basic Vertex and Fragment Shader Program
# Compatible with Jupyter Notebook / Google Colab: renders a triangle with Matplotlib.
# The GLSL shader source is included below for the experiment record.

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

# Vertex shader (GLSL)
vertex_shader_source = """
#version 330 core
layout (location = 0) in vec2 aPosition;
void main()
{
    gl_Position = vec4(aPosition, 0.0, 1.0);
}
"""

# Fragment shader (GLSL)
fragment_shader_source = """
#version 330 core
out vec4 FragColor;
void main()
{
    FragColor = vec4(0.15, 0.55, 0.95, 1.0);
}
"""

# Three vertices in normalized device coordinates
vertices = np.array([
    [-0.75, -0.60],
    [ 0.75, -0.60],
    [ 0.00,  0.75]
])

# Simulate the vertex shader's position output and fragment shader's color output
vertex_positions = vertices.copy()
fragment_color = (0.15, 0.55, 0.95, 1.0)

fig, ax = plt.subplots(figsize=(7, 6))
triangle = Polygon(
    vertex_positions,
    closed=True,
    facecolor=fragment_color,
    edgecolor="black",
    linewidth=1.5
)
ax.add_patch(triangle)

ax.set_xlim(-1, 1)
ax.set_ylim(-1, 1)
ax.set_aspect("equal", adjustable="box")
ax.set_title("Basic Shader Program: Colored Triangle")
ax.set_xlabel("X coordinate")
ax.set_ylabel("Y coordinate")
ax.grid(True, linestyle="--", alpha=0.4)
ax.axhline(0, linewidth=0.8)
ax.axvline(0, linewidth=0.8)
plt.show()

print("Vertex shader source:")
print(vertex_shader_source)
print("Fragment shader source:")
print(fragment_shader_source)
print("Triangle rendered successfully.")
```
